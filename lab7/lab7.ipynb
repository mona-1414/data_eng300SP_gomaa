MongoDB is a popular NoSQL database that stores data in JSON-like documents, making it flexible and scalable. Python, with its rich ecosystem of libraries, provides excellent support for interacting with `MongoDB`. In this tutorial, we'll go through the basics of using MongoDB with Python using the pymongo library.

### 
- **MongoDB**: Stores data as BSON (Binary JSON) documents inside collections (analogous to tables in relational databases). This structure provides flexibility since each document can have different fields.
- **SQL Databases**: Use a structured, tabular format with predefined schemas. Data is stored in rows and columns with strict relationships.

### Query Language
- **MongoDB**: Uses a JSON-like query language with methods like .find(), .insert(), .update(), etc.
- **SQL Databases**: Use Structured Query Language (SQL) with SELECT, INSERT, UPDATE, DELETE.

### MongoDB:
- **MongoDB**:  Initially, MongoDB lacked full ACID (Atomicity, Consistency, Isolation, Durability) transactions.
Modern MongoDB (v4.0+) supports ACID transactions across multiple documents but is still not as strong as SQL databases.
- **SQL Databases**: Naturally ACID-compliant, ensuring reliable, consistent transactions.

### Scalability & Performance
- **MongoDB**:  Horizontally scalable using sharding—dividing data across multiple servers.
- **SQL Databases**: Typically vertically scalable—requiring larger, more powerful servers.

### Joins & Relationships
- **MongoDB**:  Does not support joins like SQL. Instead, it uses embedded documents and references.
- **SQL Databases**: Use foreign keys and joins to relate tables efficiently.

Choose **MongoDB** if:
- Your application requires high availability and scalability (e.g., distributed cloud apps, IoT).
- You need flexible schemas and do not require strong relational constraints.
- You can tolerate eventual consistency in exchange for speed.

  
Choose **SQL Databases** if:
- You need strict consistency and full ACID compliance (e.g., financial transactions).
- Your application requires complex relationships and joins.
- You need highly structured and normalized data.

then, import `pymongo` and establish a connection to a local MongoDB instance.

````python
import pymongo

# Connect to MongoDB server running on localhost at the default port (27017)
client = pymongo.MongoClient("mongodb://localhost:27017/")

# List databases
print(client.list_database_names())
````

## 3. Creating a Database and Collection
MongoDB organizes data into databases and collections (similar to tables in SQL databases). Note that MongoDB does not require explicit database creation. A database and collection will be created when data is first inserted.

````python
# Create or access a database
mydb = client["mydatabase1"]

# Create or access a collection (table)
mycollection = mydb["customers"]
````

## 4. Inserting Documents
MongoDB stores data as BSON (Binary JSON). You can insert single or multiple documents as follows:
### Insert One Document

````python
data = {"name": "Alice", "age": 25, "city": "New York"}
result = mycollection.insert_one(data)
print("Inserted ID:", result.inserted_id)
````

### Insert Multiple Documents

````python
data_list = [
    {"name": "Bob", "age": 30, "city": "Los Angeles"},
    {"name": "Charlie", "age": 35, "city": "Chicago"}
]
result = mycollection.insert_many(data_list)
print("Inserted IDs:", result.inserted_ids)
````

### Load Data from JSON File
If you have a JSON file containing multiple records, you can load and insert them into MongoDB:

````python
import json

# Load JSON file
def load_json(filename):
    with open(filename, "r") as file:
        data = json.load(file)
    return data

json_data = load_json("data/data.json")  # Ensure data.json is a list of dictionaries

# Insert JSON data into MongoDB
result = mycollection.insert_many(json_data)
print("Inserted IDs:", result.inserted_ids)
````

### Load Data from CSV File
To insert data from a CSV file, first convert it to a list of dictionaries:

````python
import pandas as pd

# Load CSV file
def load_csv(filename):
    df = pd.read_csv(filename)
    return df.to_dict(orient="records")  # Convert DataFrame to list of dictionaries

csv_data = load_csv("data/data.csv")

# Insert CSV data into MongoDB
result = mycollection.insert_many(csv_data)
print("Inserted IDs:", result.inserted_ids)
````

## 5. Querying Data
To retrieve data from MongoDB, use `find_one()` or `find()`

### Find One Document

````python
doc = mycollection.find_one()
print(doc)
````

### Find Multiple Documents

````python
for doc in mycollection.find():
    print(doc)
````

You can also specify query conditions:

````python
query = {"age": {"$gt": 28}}  # Find documents where age > 28
for doc in mycollection.find(query):
    print(doc)
````

## 6. Updating Documents
To update a document, use `update_one()` or `update_many()`.

````python
query = {"name": "Alice"}
new_values = {"$set": {"age": 26}}
mycollection.update_one(query, new_values)
````

Note: if there are multiple rows with filed `name` as 'Alice'.

## 7. Deleting Documents
To delete documents, use `delete_one()` or `delete_many()`.

````python
query = {"name": "Charlie"}
mycollection.delete_one(query)
query = {"age": {"$lt": 30}}  # Delete all documents where age < 30
mycollection.delete_many(query)
````

## 8. Indexing for Performance
Indexes improve query performance. To create an index and list the index

````python
mycollection.create_index("name")
print(mycollection.index_information())
````

## 9. Closing the Connection
Always close the connection when done:

````python
client.close()
````
